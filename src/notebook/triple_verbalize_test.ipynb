```python
# Setting system path and project root
import os
import sys

PROJECT_ROOT_DIR = os.path.abspath('../../')
sys.path.append(PROJECT_ROOT_DIR) # bringing system path to project root

def get_fp(relative_path):
    return os.path.join(PROJECT_ROOT_DIR, relative_path)
```

```python
# Load the entity map
import pickle
from pathlib import Path

def load_label_map(pickle_path: str | Path) -> dict:
    """Return the subject -> labels mapping saved by label_map_gen.py."""
    path = Path(pickle_path)
    if not path.is_file():
        raise FileNotFoundError(f"No such file: {path}")

    with path.open("rb") as f:
        mapping = pickle.load(f)  # type: dict
    return mapping
```

```python
label_file_path = get_fp("data_dir/verbalization/label_map.pkl")
label_map = load_label_map(label_file_path)
print('Label map loaded!')
```

```python
import random
def print_random_entries(label_map: dict, n: int = 10):
    total = len(label_map)
    if total == 0:
        print("Mapping is empty.")
        return

    sample_keys = random.sample(list(label_map.keys()), min(n, total))
    print(f"Showing {len(sample_keys)} random entries (out of {total:,} subjects):\n")
    for key in sample_keys:
        labels = label_map[key]
        # Show only the first few labels to keep output readable
        preview = labels[:3]
        printable = "[" + ", ".join(f"'{l}' - {len(l)}" for l in preview) + "]"
        more = "…" if len(labels) > 3 else ""
        print(f"{key}\n    : {printable}{more}  ({len(labels)} label(s))\n")
```

```python
print_random_entries(label_map, n=10)
```
